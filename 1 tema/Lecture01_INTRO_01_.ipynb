An if statement can be optionally followed by one or more elif blocks and a catch- all else block if all of the conditions are False

````python
x = 4
if x < 0:
    print('It\'s negative')
elif x == 0:
    print('Equal to zero')
elif 0 < x < 5:
    print('Positive but smaller than 5')
else:
    print('Positive and larger than or equal to 5')
````

````python
a = 5; b = 7
c = 8; d = 4
if a < b or c > d:
    print('Made it')
````

#### CONTINUE, PASS, BREAK

````python
sequence = [1, 2, None, 4, None, 5]
total = 0
for value in sequence:
    if value is None:
        continue
    total += value
total
````

A for loop can be exited altogether with the break keyword. This code sums elements of the list until a 5 is reached:

````python
sequence = [1, 2, 0, 4, 6, 5, 2, 1]
total_until_5 = 0
for value in sequence:
    if value == 5:
        break
    total_until_5 += value
total_until_5
````

The break keyword only terminates the innermost for loop; any outer for loops will
continue to run:

````python
for i in range(4):
    for j in range(4):
        if j > i:
            break
        print((i, j))
````

#### While Loops

A while loop specifies a condition and a block of code that is to be executed until the condition evaluates to False or the loop is explicitly ended with break:


````python
x = 256
total = 0
while x > 0:
    if total > 500:
        break
    total += x
    x = x // 2
````
