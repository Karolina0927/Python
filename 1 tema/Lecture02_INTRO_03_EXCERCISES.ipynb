### EX. 5

use dictionary to put word according to their first letter into a dictionary values words = ['apple', 'bat', 'bar', 'atom', 'book']

```python
box = {}
words = ['apple', 'bat', 'bar', 'atom', 'book']
for word in words:
    letter = word[0]
    if letter not in box.keys():
        box[letter] = [word]
    else:
        box[letter].append(word)
box 
```

### EX. 9

Use zip and enumerate for two list and print it: 
seq1 = ['foo', 'bar', 'baz']
seq2 = ['one', 'two', 'three']

```python
seq1 = ['foo', 'bar', 'baz']
seq2 = ['one', 'two', 'three']
for i, (a, b) in enumerate(zip(seq1, seq2)):
    print('{0}: {1}, {2}'.format(i, a, b))

```

### EX. 11

Write a code that filters names that have a letter all_data = [['John', 'Emily', 'Michael', 'Mary', 'Steven'],['Maria', 'Juan', 'Javier', 'Natalia', 'Pilar']

```python
all_data = [['John', 'Emily', 'Michael', 'Mary', 'Steven'],['Maria', 'Juan', 'Javier', 'Natalia', 'Pilar']]
names_of_interest = []
for names in all_data:
    enough_es = [name for name in names if name.count('a') >= 1]
    names_of_interest.extend(enough_es)
names_of_interest
```
