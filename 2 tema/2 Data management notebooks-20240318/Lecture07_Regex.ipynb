Regular expressions provide a flexible way to search or match (often more complex) string patterns in text. A single expression, commonly called a regex, is a string formed according to the regular expression language. Python’s built-in re module is responsible for applying regular expressions to strings; I’ll give a number of examples of its use here.

The re module functions fall into three categories: pattern matching, substitution, and splitting. Naturally these are all related; a regex describes a pattern to locate in the text, which can then be used for many purposes. Let’s look at a simple example: suppose we wanted to split a string with a variable number of whitespace characters (tabs, spaces, and newlines).

The following describes regular expressions’ functionalities.
- Find a single occurrence of characters a and b: [ab]

```python
import re
text = "foo bar\t baz \tqux"
re.findall("[ab]", text)
```

- Find characters except for a and b: [^ab]

```python
import re
text = "foo bar\t baz ,,\tx99"
re.findall("[^ab]", text)
```

- Find groups of a and b: [^ab]+

```python
import re
text = "foo bar\t baaz \tqux"
re.findall("[ab]+", text)
```

- Find the character range of a to z: [a-z]

```python
import re
text = "foo bar\t baz \tqux"
re.findall("[a-z]", text)
```

- Find a character range except a to z: [^a-z]

```python
import re
text = "foo bar\t baz \tqux"
re.findall("[^a-z]", text)
```

- Find all the characters from both a to z and A to Z: [a-zA-Z]

```python
import re
text = "foo bar\t bZaz \tquxZ"
re.findall("[a-zA-Z]", text)
```

- Find any whitespace character: \s

```python
import re
text = "foo bar baz qux"
re.findall("\s", text)
```

- Find any non-whitespace character: \S

```python
import re
text = "foo bar baz qux?!"
re.findall("\S", text)
```

- Find any digit: \d

```python
import re
text = "price 24 eur"
print(re.findall("\d", text))
print(re.findall("\d+", text))
```

- Find any non-digit: \D

```python
import re
text = "foo bar ba2z q3!"
re.findall("\D+", text)
```

- Find any non-words: \W

```python
import re
text = "foo bar ba2z! q3ux?"
re.findall("\W", text)
```

- Find any words: \w

```python
import re
text = "foo bar ba2z! q3ux?"
re.findall("\w", text)
```

Parentheses () are used to group parts of a regular expression together. This can be useful for applying quantifiers, such as *, +, or ?, to a group of characters, or for capturing a specific substring of the matched text. For example, the regular expression pattern (ab)+ would match one or more occurrences of the substring "ab" in a string, and capture the last occurrence of "ab" in the matched text.

Brackets [] are used to define a character class, which matches any single character that is contained within the brackets. For example, the regular expression pattern [abc] would match any single occurrence of the characters "a", "b", or "c". Brackets can also be used to define a range of characters, such as [a-z] which would match any lowercase letter between "a" and "z", inclusive.

- Find either a or b: (a|b)

```python
import re
text = "foo baraabcd ba2z! q3ux?"
re.findall("(ab)", text)
```

```python
re.findall("[ab]+", text)
```

- Zero or one occurrences a?. It is like a dummy variable

```python
import re
text = "firl aaraaogueaaa"
re.findall("a?", text)
```

- Zero or more occurrences a*. It is like a dummy variable

```python
import re
text = "freba aaraaueaaa"
re.findall("a*", text)
```

- the occurrence of a is one or more times: a+

```python
import re
text = "fireball ambaraaogueaaa"
re.findall("a+", text)
```

- match three simultaneous occurrences of a: a{2}

```python
import re
text = "fireball ambaraaogueaaa"
re.findall("a{2}", text)
```

- Match two or more simultaneous occurrences of a: a{2,}

```python
import re
text = "fireball ambaraaogueaaa"
re.findall("a{2,}", text)
```

- Match three to six simultaneous occurrences of a: a{3,6}

```python
import re
text = "fireball ambaraaogueaaa gaaaalaaaaa"
re.findall("a{3,6}", text)
```

 - Start of a string: ^. Checks if the sentence starts with a certain word

```python
import re
text = "fireball ambaraaogueaaa gaaaalaaaaa"
re.findall("^fireball", text)
```

 - end of a string: $

```python
import re
text = "fireball ambaraaogueaaa gaaaalaaaaa"
re.findall("gaaaalaaaaa$", text)
```

Returns a match where the specified characters are at the beginning (the "r" in the beginning is making sure that the string is being treated as a "raw string")

```python
import re
text = "fireball fambaraaogueaaa gaaaalaaaaaf"
re.findall(r"\bg", text)
```

Returns a match where the specified characters are present, but NOT at the beginning (or at the end) of a word
(the "r" in the beginning is making sure that the string is being treated as a "raw string")

```python
import re
text = "firebll ambaraaaoguea gl"
print(re.findall(r"\Ba", text)) #not at the beggining
### if you want not at the end just move \B at the end likes this:
re.findall(r"a\B", text)
```

### REGEX FUNCTIONS

- findall - Returns a list containing all matches
- search - Returns a Match object if there is a match anywhere in the string
- split	- Returns a list where the string has been split at each match
- sub - Replaces one or many matches with a string

```python
import re
txt = "The rain in Spain"
re.search("\s", txt)
```

```python
txt = "The rain in Spain"
x = re.search("rain", txt)
print(x)
```

```python
txt = "The rain in Spain"
x = re.split("\s", txt)
print(x)
```

```python
txt = "The rain in Spain"
x = re.sub("\s", "9", txt)
print(x)
```

#### Extracting Email IDs

```python
doc = "For more +details please mail us at: xyz@abc.com, pqr@mno.com"
addresses = re.findall(r'[\w\.-]+@[\w\.-]+', doc) 
for address in addresses:
    print(address)
```
