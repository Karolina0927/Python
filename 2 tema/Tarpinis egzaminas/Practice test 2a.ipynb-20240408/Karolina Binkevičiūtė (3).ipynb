### Q1

1. Create a dataframe with 1020 rows and twenty columns. Values can be random.
2. Change the names of the columns into Lithuanian city names.
3. Calculate the means of top three biggest cities in your dataset.
4. Remove the first city column.
5. For each column calculate it's mean and change each column's values to the mean.

```python
import numpy as np
import pandas as pd
```

```python
data = pd.DataFrame(np.arange(20400).reshape((1020, 20)))
data
```

```python
data = pd.DataFrame(np.arange(20400).reshape((1020, 20)),
                   columns=['Birštonas', 'Kaunas', 'Vilnius', 'Klaipėda', 'Biržai', 'Druskininkai', 'Prienai', 'Šiauliai', 'Alytus', 'Jonava', 'Kėdainiai', 'Kretinga', 'Kupiškis', 'Palanga', 'Šventoji', 'Marijampolė', 'Pakruojis', 'Panevėžys','Plungė' ,'Rokiškis'])
data
```

```python
means = data[['Kaunas', 'Vilnius', 'Klaipėda']].mean()
print(means)
```

```python
no_first_column = data.drop(columns=['Birštonas'])
no_first_column
```

```python
mean = no_first_column.mean()
mean
```

```python
for column in data.columns:
    data[column] = data[column].mean()
data
```

4. Calcualte cross-correlation between the companies. 
5. If I can only choose two companies out of five which ones should I choose in order to have a diverse portfolio?

```python
import yfinance as yf
```

```python
ticker_symbols = ['AAPL', 'MSFT', 'AMZN', 'META', 'GOOGL']
stock_data = yf.download(ticker_symbols, start='2023-01-01', end='2023-12-31')
stock_data
```

```python
stock_data.to_csv('top_five_tech_companies_stock_data.csv')
```

```python
last_day_close = stock_data.iloc[-1]['Close']
company_highest_price = last_day_close.idxmax()
highest_price = last_day_close.max()

print("Company:", company_highest_price)
print("Stock Price:", highest_price)
```

```python
close_prices = stock_data['Close']
correlation_matrix = close_prices.corr()
print(correlation_matrix)
```
