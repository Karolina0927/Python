### EX. 14

Create the dataframe below. Create a lamba funciton that retunrs minimum and maximum values in a series.

```python
frame2 = pd.DataFrame(np.arange(0,20.).reshape(4, 5), columns=['a', 'b', 'c', 'd', 'e'], index=['0','1', '2', '3'])
frame2
```

```python
def f(x):
    return pd.Series([x.min(), x.max()], index=['min', 'max'])
frame2.apply(f)
```
