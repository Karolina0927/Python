### Correlation
- Pearson Correlation: measures the linear relationship between two variables X and Y. It assumes that both variables are normally distributed and have a linear relationship between them. The approximately normal distirbution is required if you want to tst using p-values.

$$ r=  \frac{\Sigma(x_i - \bar{x})(y_i - \bar{y}) }{\sqrt{\Sigma(x_i-\bar{x})^2\Sigma(y_i-\bar{y})^2}}  \hspace{0.5in}  \rho=\frac{Cov(x,y)}{\sigma_x\sigma_y} $$

```python
print("""0.8 - 1: very strong,
0.6 - 0.79: strong,
0.4 - 0.59: moderate,
0.2 - 0.39: weak
0 - 0.19: very weak""")
```

Is there a difference between sample and population correlations?. But do we get the same correlations? Yes for both. The denominator for Covariance is either n or 'n - 1' for population and sample, respectively. To calculate the correlation for a sample, it is the covariance divided by SxSy (which each are the square root of their variance, that has 'n - 1' in the denominator). To calculate the correlation for a population, it is the covariance divided by Sigma X and Sigma Y (which each are the square root of their variance, that has 'n' in the denominator). Thus, either the 'n - 1' cancels out when calculating the correlation for a sample, or the 'n' cancels out when calculating the correlation for a population.

#### Calculate Pearson correlation using the formula

 $$r=  \frac{\Sigma(x_i - \bar{x})(y_i - \bar{y}) }{\sqrt{\Sigma(x_i-\bar{x})^2\Sigma(y_i-\bar{y})^2}}  \hspace{0.5in}  \rho=\frac{Cov(x,y)}{\sigma_x\sigma_y} $$

### Kendall Correlation: 
- measures the ordinal association between two variables X and Y. It is also based on rank-order correlations and does not assume normality.

$$ \tau = \frac{n_c - n_d}{n(n-1)/2} \hspace{0.1in} \textrm{where} \hspace{0.1in}  n_c \hspace{0.1in} \textrm{is concordant pairs and}\hspace{0.1in} n_d \hspace{0.1in} \textrm{discordant pairs}$$

```python
import numpy as np
from scipy.stats import kendalltau

# Generate x data with a pattern
n = 100
x = np.arange(n)
y = np.random.rand(n)

# Add a sine wave pattern to y data
y += np.sin(0.1 * x)

# Calculate Kendall's correlation coefficient
tau, p_value = kendalltau(x, y)

print("Kendall's tau:", tau)
print("p-value:", p_value)
plt.scatter(x,y)
```

### Distance correlation:
- Distance correlation or distance covariance is a measure of dependence between two paired random vectors of arbitrary, not necessarily equal, dimension. The population distance correlation coefficient is zero if the random vectors are independent. Thus, distance correlation measures both linear and nonlinear association between two random variables or random vectors.
- Instead of assessing how two variables tend to co-vary in their distance from their respective means, distance correlation assesses how they tend to co-vary in terms of their distances from all other points.

$$ dCor(X,Y) = \frac{dCov(X,Y)}{\sqrt{dVar(X)dVar(Y)}} $$

#### Calculate distance correlation  using the following formulas

Let (Xk, Yk), k = 1, 2, …, n be a statistical sample from a pair of two random variables, X & Y.

```python
import numpy as np
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 1, 5, 3])
```

First, we compute the n by n distance matrices (aj, k) and (bj, k) containing all pairwise distances.

$$ a_{j,k} = ||X_j - X_k|| \hspace{0.1in} j,k = 1,2...,n$$
$$ b_{j,k} = ||Y_j - Y_k|| \hspace{0.1in} j,k = 1,2...,n$$

Then we take the double centered distances.

$$ A_{j,k} = a_{j,k} - \bar{a_j} - \bar{a_k} + \bar{a}....$$
$$ B_{j,k} = b_{j,k} - \bar{b_j} - \bar{b_k} + \bar{b}....$$

Lastly, we compute the arithmetic average of the products A and B to get the squared sample distance covariance. The distance variance is simply the distance covariance of two identical variables. It is the square root of the following:

$$ dCov^2(X,Y) = \frac{1}{n^2}  \Sigma^n_{j=1}\Sigma^n_{k=1}A_{j,k}B_{j,k} $$
$$ dVar^2_n(X) = dCov^2_n(X,X) = \frac{1}{n^2}\Sigma_{k,l}A^2_{k,l}$$

From a visual perspective, by taking the double centered distances, we are transforming the matrix representation (the left) to the diagram on the right (double centered matrix).

### Cosine similarity:
Cosine similarity​ is a metric often used to measure how similar the documents are irrespective of their size. Mathematically, it measures the cosine of the angle between two vectors projected in a multi-dimensional space. The cosine similarity is advantageous because even if the two similar documents are far apart by the Euclidean distance (due to the size of the document), chances are they may still be oriented closer together. The smaller the angle, higher the cosine similarity.

- Uses the angle between 2 lines.
- When the angle is  0  (the vectors point in the exact same direction), 𝑐𝑜𝑠(𝜃) will equal 1.
- When the angle is  −180°, (the vectors point in exact opposite directions), then  𝑐𝑜𝑠(𝜃) will equal  −1.
- When the angle is  90° (the vectors point in completely unrelated directions), then  𝑐𝑜𝑠(𝜃)  will equal 0.

$$  \overrightarrow{a} \cdot \overrightarrow{b} = ||\overrightarrow{a}||||\overrightarrow{b}||cos\theta$$
$$cos\theta= \frac{\overrightarrow{a}\cdot\overrightarrow{b}}{||\overrightarrow{a}||||\overrightarrow{b}||}$$

#### Calculate cosine similarity using the formula

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Define the x and y arrays
x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])

# Calculate the cosine similarity using Scikit-learn
x = x.reshape(1, -1) # Reshape to a 2D array
y = y.reshape(1, -1) # Reshape to a 2D array
cosine_similarity_sklearn = cosine_similarity(x, y)
print("Cosine similarity using Scikit-learn:", cosine_similarity_sklearn[0][0])
```

### Mutual Information
The mutual information (MI) between a feature and the outcome is a measure of the mutual dependence between the two variables. It is given the name ,information gain, when applied to variable selection. It extends the notion of correlation to nonlinear relationships. More specifically, it quantifies the information obtained about one random variable through the other random variable. MI determines how different the joint distribution of the pair ( 𝑋
 , 𝑌) is to the product of the marginal distributions of 𝑋 and 𝑌.

- Let  (𝑋,𝑌) be a pair of random variables with values over the space [] ×[]. If their joint distribution is  𝑃(𝑋,𝑌) and the marginal distributions are  𝑃𝑋 and  𝑃𝑌
, the mutual information is defined as

$$  MI(X;Y) = D_{KL}(P_{(X,Y)}||P_X \otimes P_Y)$$
$$ MI = \Sigma_{x\in X }\Sigma_{y \in Y}p(x,y)log \frac{p(x,y)}{p(x)p(y)} $$

### Maximal information Coefficient
- The maximal information coefficient (MIC) (Simon and Tibshirani,  2011) is an information theory-based measure of association that can capture a wide range of functional and non-functional relationships between variables.
- Formally, MIC is equal to the coefficient of determination (R2). MIC takes values between 0 and  1, where 0 means statistical independence and  1
means a completely noiseless relationship.

$$ MIC(X,Y) = max\{I(x,y)\} / log_2 min\{n_x, n_y\}$$

- From the formula we can see that MIC(𝑋, 𝑌) is the mutual information between random variables 𝑋 and 𝑌 normalized by their minimum joint entropy. I interpret the MIC as the percent of a variable  𝑌
that can be explained by a variable  𝑋.
- In addition to generalizing well over a range of relationships, another useful property of MIC is its equitability. This means that MIC assigns the same score to equally noisy relationships, regardless of the type of relationship. This is good because a lot of times we do not know the distribution of our data or the nature of relationships between variables.
- Able to capture wide range of linear and non-linear relationships (cubic, exponential, sinusoidal, superposition of functions)
- Does not make any assumptions about the distribution of the variables.
- Robust to outliers because of its mutual information foundation.
- Does not report the direction or the type of relationship.
- Low statistical power: when Simon and Tibshirani compared the performance of MIC with Pearson’s and distance correlation on simulated independent data, they reported that in most cases MIC had less power.

```python
import minepy

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

mine = minepy.MINE(alpha=0.6, c=15, est='mic_approx')
mine.compute_score(x,y)
mic =  mine.mic() 
print("MIC value:", mic)
```

```python
ts_dict = {'lin':      {'x': x_l, 'y': y_l0}, 
           'lin_n':    {'x': x_l, 'y': y_l1},
           'exp lin':  {'x': x_e, 'y': y_e0}, 
           'exp lin_n':{'x': x_e, 'y': y_e1},
           'quadr':    {'x': x_q, 'y': y_q0}, 
           'quadr_n':  {'x': x_q, 'y': y_q1},
           'sin':      {'x': x_l, 'y': y_s0}, 
           'sin_n':    {'x': x_l, 'y': y_s1},
           'cir':      {'x': x_c,  'y': y_c}, 
           'cir_n':    {'x': x_cn, 'y': y_cn}}
```

```python
def compare_corrs(Dict=ts_dict):
    
    mine = MINE(alpha=0.6, c=15, est='mic_approx')
    dist_corrs, sprearman_corrs, pearson_corrs, mi_reg, f_tests, cos_sims, mic_s = [], [], [], [], [], [], []
    
    for ts in Dict.keys():
    
        X = ts_dict[ts]['x']
        y = ts_dict[ts]['y']

        dist_corr = scipy.spatial.distance.correlation(X,y)
        sprearman_corr, s_p_val = stats.spearmanr(X,y)
        pearson_corr, p_p_val = stats.pearsonr(X,y)
        cos_dist = scipy.spatial.distance.cosine(X,y)
    
        mi = mutual_info_regression(X.reshape(-1,1), y.reshape(-1,1))
        f_test, _ = f_regression(X.reshape(-1,1), y.reshape(-1,1))
    
        mine.compute_score(X,y)
        mic =  mine.mic() 
    
        dist_corrs.append(dist_corr)
        sprearman_corrs.append(sprearman_corr)
        pearson_corrs.append(pearson_corr)
        cos_sims.append(1-cos_dist)
        mi_reg.append(mi[0])
        f_tests.append(f_test[0])
        mic_s.append(mic)
    
    corrs_df = pd.DataFrame(dist_corrs, index=ts_dict.keys(), columns=['Dist_corr'])
    corrs_df['cos_sim'] = cos_sims
    #corrs_df['f_reg'] = f_tests / np.max(f_tests)
    corrs_df['mutual_info'] = mi_reg / np.max(mi_reg)
    corrs_df['spear_corr'] = sprearman_corrs
    corrs_df['pear_corr'] = pearson_corrs
    corrs_df['mic'] = mic_s

    return corrs_df
```

#### Some notes on why n-1 exists when calculating variance but not the mean.

Imagine a small population, it has 3 numbers. 1 & 2 & 3.Now, get 2 samples (with replacement) from that population, and get all the possible samples you could get. You will get:
- (1,1)
(1,2)
(1,3)
(2,1)
(2,2)
(2,3)
(3,1)
(3,2)
(3,3)

Some samples represent the population better than others, but we want to understand how well, overall, samples represent the population.
The mean of the population is 2. Get the mean of each sample Sum (X) / N, and make a histogram. (i.e., (1+1)/2 = 1, 2+1 / 2 = 1.5)
It should look like the one below.

```python
import matplotlib.pyplot as plt

data = [1.5, 1.5, 2.0, 2.0, 2.0, 2.5, 2.5, 3.0]

plt.hist(data, bins=6, color='blue')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Data')
plt.show()
```

Pretty! The distribution of all the possible sample means is symmetric (in this example, and when N is about 30 or greater regardless), and the mean of all those means is…..2. It is equal to the mean of the population. The sample mean is not biased- On average it comes out equal to the population mean and there are roughly as many ways it can come out too small as it can come out too big.

Now, look at the Variance. First, calculate the variance of the population. You have all 3 numbers of the population, so you are calculating the actual population variance, so you would use N instead of N-1 (which you use when you are using a sample to Estimate the population variance).

```python
print(f"Population variance: {((1 -2)**2 + (2-2)**2 + (3-2)**2) / (3 )}")
```

So, how well do your sample variances estimate that value? First, calculate the variance for each sample using N and make a histogram. It should look like…

```python
0.00, 0.00, 0.00
0.25, 0.25, 0.25, 0.25
..........
1.00, 1.00
```

Calculate the mean of all those sample variances. You should get 0.333. On average the sample variance, when calculated with /N, is smaller than the variance of the population. Notice also that the distribution is skewed. There are more ways to select a sample variance that is too small than one too big, so sample variances are biased, i.e., likely to be too small.

You can correct the long term average problem by dividing by n-1 instead of n. Then, your distribution would look like

```python
0.00, 0.00
........
0.50, 0.50, 0.50, 0.50
.......
2.0, 2.0
```

Now get the average, and you should find it is .667. Using the degrees of freedom will make the average sample variance equal to the population variance. N-1 removes part of the bias. But, because the variances are squared numbers the distribution is going to be skewed (little number squared = bigger number, big number squared = huge number). There are still more ways to get a small sample variance than a big one (which is why things like the t-distribution were formulated… but that’s another story).
