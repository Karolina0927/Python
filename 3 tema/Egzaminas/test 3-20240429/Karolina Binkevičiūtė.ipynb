### Q5

- Find the third to last business day of each quater for the following date ranges: 2008 - 2024

```python
import pandas as pd

date_range = pd.date_range(start='2008-01-01', end='2024-12-31', freq='B')

df = pd.DataFrame(date_range, columns=['Date'])

df['Year'] = df['Date'].dt.year
df['Quarter'] = df['Date'].dt.to_period('Q')

thirdToLast = df.groupby(['Year', 'Quarter']).apply(lambda x: x.iloc[-3])

thirdToLast.reset_index(drop=True, inplace=True)

print(thirdToLast['Date'])
```
