2. Correlation vs causation:
- a) What is the difference between correlation and causation? 
- b) Give me an example there corelation exist but there is no causal relationship.

a) Difference Between Correlation and Causation
Correlation: Indicates a statistical relationship between two variables, meaning they change together, but does not imply that one causes the other.
Causation: Means that one event directly causes another event.
b) Example of Correlation Without Causation
Ice Cream Sales and Drowning Incidents: Both increase in the summer, showing a positive correlation. However, eating ice cream does not cause drowning. The real cause is hot weather, which leads to more people buying ice cream and swimming.

3. Implement the following formula for the following dataset:
    $$ \frac{\Sigma y_i}{\bar y} \times \frac{\Sigma x_i * \bar x}{\sigma_x}  $$
    $$ x = [10, 34, 15, 28, 34, 88, 29]$$
    $$ y = [1, 2, 3, 4, 5, 6, 7]$$

```python
x = np.array([10, 34, 15, 28, 34, 88, 29])
y = np.array([1, 2, 3, 4, 5, 6, 7])
data = (np.sum(y) / np.mean(y)) * (np.sum(x) * np.mean(x) / np.std(x))
print(data)
```

- b) Show which type of product was sold the most.
- c) Calculate how much revenue did this company made.

- e) Show at least three products that women prefer more than men.


```python
from sympy import Symbol, Derivative
from sympy import Symbol
import matplotlib
import numpy as np
import pandas as pd
```

```python
orders = pd.read_csv('orders.csv')
print(orders)
```

```python
orders[['Product','Quantity']]
```

```python
prices = pd.read_csv('prices.csv')
print(prices)
```

```python
product = orders['Product'].value_counts().idxmax()
product
```

```python
merged = pd.merge(orders,prices, left_on='Product', right_on='Fruit')
merged
```

```python
merged['Revenue'] = merged['Price'] * merged['Quantity'] 
merged
```

```python
total_revenue = merged.groupby('Product')['Revenue'].sum().sum()
print(total_revenue)
```

```python
highest_revenue = merged.groupby('Product')['Revenue'].sum().sort_values(ascending=False).head(1)
highest_revenue
```

```python
product_sex_preferences = merged.groupby(['Product', 'Sex'])['Quantity'].sum().unstack()
product_sex_preferences['Difference'] = product_sex_preferences['Female'] - product_sex_preferences['Male']
products_preferred_by_women = product_sex_preferences[product_sex_preferences['Difference'] > 0]
top_three_products_preferred_by_women = products_preferred_by_women.sort_values(by='Difference', ascending=False).head(3)

top_three_products_preferred_by_women
```
