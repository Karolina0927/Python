```python
correlation1 = df['LSTAT'].corr(df['MEDV'], method='pearson')
correlation1
print(f"Correlation between LSTAT and MEDV: {correlation1}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the matplotlib figure
plt.figure(figsize=(10, 6))

# Create a scatter plot
sns.scatterplot(x='LSTAT', y='MEDV', data=df, hue='LSTAT', palette='viridis')

# Add the regression line
sns.regplot(x='LSTAT', y='MEDV', data=df, scatter=False, color='red')

# Title and labels
plt.title('Correlation between LSTAT and MEDV')
plt.xlabel('% Lower Status of the Population (LSTAT)')
plt.ylabel('Median Value of Homes (MEDV)')

# Show the plot
plt.show()
```

```python
import pandas as pd
import numpy as np

# Round up PTRATIO values
df['PTRATIO'] = np.ceil(df['PTRATIO'])

# Define custom bin edges for PTRATIO
bin_edges = [13, 17, 19, 20, df['PTRATIO'].max() + 1]


# Create custom groups for PTRATIO
df['PTRATIO_Group'] = pd.cut(df['PTRATIO'], bins=bin_edges, labels=['(13, 17]', '(17, 19]', '(19, 20]', '(20, 23]'], include_lowest=True)

# Calculate the mean MEDV for each PTRATIO group
pivot_table = df.pivot_table(values='MEDV', index='PTRATIO_Group', aggfunc='mean')
pivot_table['MEDV'] = pivot_table['MEDV'].round(2)


# Print the pivot table
pivot_table
```
